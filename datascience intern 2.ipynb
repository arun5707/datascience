```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('C:\\Users\\ARUN\\Downloads\\apple_jobs_cleaned.csv')

print(df.describe(include='all'))
print(df.isnull().sum())

plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='state_region', order=df['state_region'].value_counts().index[:10])
plt.title('Top 10 Regions for Apple Jobs')
plt.show()

top_titles = df['title'].value_counts().head(10)
plt.figure(figsize=(10, 5))
top_titles.plot(kind='pie', autopct='%1.1f%%')
plt.title('Top 10 Most Frequent Job Titles')
plt.ylabel('')
plt.show()
```
